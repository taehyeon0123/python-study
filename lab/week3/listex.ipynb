```python
student = 10
ist = []
count = 0

for i in range(student):
    value = int(input("학생의 성적를 입력하세요: "))
    lst.append(value)

print("\n성적 평균=", sum(lst) / len(lst))
print("성적 최대값=", max(lst))
print("성적 최소값=", min(lst))

for score in lst:
    if score >= 80:
        count += 1
    print("80점 이상 학생 수=", count)
```
