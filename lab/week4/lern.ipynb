# counter 클래스 만들기

1. 데이터 , 전체 숫자
2. 동작

```python
class Counter :
    def __init__(self, initial_count=0): # 생성함수
        self.count = initial_count
    
    def increment(self, amount = 1):
        self.count += amount

    def get_count(self):
        return self.count


def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for _ in range(10):
        myc.increment()
    print(myc.get_count())
   
    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment()
    print(yourc.get_count())
    yourc.increment(4)
    print(yourc.get_count())
    
    
if __name__ == "__main__":
    test_counter()
    
```

TV 클래스 추가

데이터: 현재 채널번호, 볼륨량, ON 여부

동작: 켜기, 끄기, 채널변경, 볼륨변경, 현재 TV 객체 내용 출력

클래스명: MyTelevision

```python

class MyTelevision:
    serialNumber = 0
    def __init__(self, channel=1, volume=10, is_on=False):
        self.channel = channel
        self.volume = volume
        self.is_on = is_on
        MyTelevision.serialNumber += 1
        self.myserial = MyTelevision.serialNumber
    
    def turn_on(self):
        self.is_on = True
    
    def turn_off(self):
        self.is_on = False

    def change_channel(self, channel):
        self.channel = channel
        
    def change_volume(self, volume):
        self.volume = volume
        
    def display_status(self):
        print(f"Channel: {self.channel}, Volume: {self.volume}, Is On: {self.is_on}")

       
        
def test_television():
    mytv = MyTelevision(channel = 11)
    mytv.turn_on()
    mytv.display_status()
    mytv.turn_off()
    mytv.display_status()
    
if __name__ == "__main__":
    test_television()
```

TV 리모콘 클래스

데이터는 없고 동작만 있음
연관된 TV를 켜거나 꺼라는 동작만 지원
TV 상태 표시 추가

```python
class TvRemocon :
    def __init__(self, tv: MyTelevision):
        self.tv = tv
        
    def press_on(self):
        self.tv.turn_on()
        
    def press_off(self):
        self.tv.turn_off()

    def press_display(self):
        self.tv.display_status()
        
def test_remocon():
    tv1 = MyTelevision()
    rc1 = TvRemocon(tv1)
    rc1.press_on()
    rc1.press_display()
    rc1.press_off()
    rc1.press_display()
    
if __name__ == "__main__":
    test_remocon()
```

# 자동차 객체 생성


```python
class Car:
    def __init__(self, speed, color, model):
        self.speed = speed
        self.color = color
        self.model = model

    def drive(self):
        self.speed = 120


c = Car(0, "black", "bmw")

print("자동차 객체를 생성하였습니다")
print("자동차 속도는", c.speed)
print("자동차 색깔은는", c.color)
print("자동차 모델은", c.model)    

c.drive()
print("bmw 속도는",c.speed)
```

가시성 처리 방법

private: 변수나 함수이름을 __로 시작한다.

외부에서 접근이 불가능하다.

```python
class Student:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def display_info(self):
        print(f"Name: {self.__name}, Age: {self.__age}")
        super().__init__(name, age)
kim = Student("Kim", 21)
kim.display_info()
print(kim.__age)  # 접근 불가
```

우리 팀의 4명에 대한 객체를 생성하고 목록에 저장 :( list 사용)
for 문을 통해 목록에 있는 모든 객체의 이름과 나이 출력

테스트 함수 수

```python
from datetime import datetime

class Person:
    def __init__(self, name, birth_year):
        self.__name = name
        self.__birth_year = birth_year

    def get_name(self):
        return self.__name

    def get_age(self):    
        current_year = self.__get_current_year()
        return current_year - self.__birth_year
    
    def __get_current_year(self):
        return datetime.now().year
    
    def __str__(self):
        return f"Name: {self.__name}, Age: {self.get_age()}"
    
def test_person() :
    song = Person("Song", 2001)
    y = song.__get_current_year()
    print(y)
    team_members = [Person("Kim", 2000), Person("Lee", 1999)]
    for member in team_members:
        print(member)
        
if __name__ == "__main__":
    test_person()   
```
