# 원의 넓이 구하기

```python
import math

class Circle:       
    def __init__(self, radius=0):
        self.radius = radius

    def getArea(self):
        return math.pi * self.radius * self.radius

    s = Circle(5)
    print(s.getArea())
```

# 자동차 객체 생성


```python
class Car:
    def __init__(self, speed, color, model):
        self.speed = speed
        self.color = color
        self.model = model

    def drive(self):
        self.speed = 120


c = Car(0, "black", "bmw")

print("자동차 객체를 생성하였습니다")
print("자동차 속도는", c.speed)
print("자동차 색깔은는", c.color)
print("자동차 모델은", c.model)    

c.drive()
print("bmw 속도는",c.speed)
```
