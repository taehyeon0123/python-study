상속에 대한 이론과 학습

1. 캡슐화 : 데이터와 연산을 하나의 덩어리로 묶어서 다룸
➢ 정보은닉을 위한 가시성 : public (공개), protected (보호), private(숨김)
◼ 다른 클래스에서 이 클래스의 데이터와 연산을 접근하거나 접근불가로 만듬
◼ Public: 모든 클래스에서 접근가능, protected: 자식만 접근가능, private : 접근 불가
2. 상속 : 부모클래스를 자식클래스가 그대로 물려받아 추가적인 코드를 작성할 수 있음
➢ 다만, 부모의 private 한 것은 상속은 받으나 접근 불가능
◼ Protected는 외부 접근 불가능하나 자식이 물려받아 사용가능
3. 다형성 : 함수 이름이 같더라도 문맥(혹은 클래스)에 따라 다른 함수가 수행됨
➢ A.do(), B.do() 처럼 do()함수 이름은 같으나 클래스가 다르면 다른 함수
➢ B가 A를 상속하고 재정의 할 경우 동적 바인딩이 가능

# private 멤버 부모 클래스

```python
class Parent(object):
    def __init__(self):
        self.__money = 100

class Child(Parent):
        def __init__(self):
            super().__init__()
            self.__money = 200
        def get_money(self):
            return self.__money

obj = Child()
print(obj.get_money())  # This will raise an AttributeError since __money is private in both classes
```

# 다중 상속

```python
class Person:
    def __init__(self, name, age):
            self.name = name
            self.age = age

    def show_info(self):
        print(self.name, self.age)

class Student:
        def __init__(self,id):
            self.id = id
        def get_id(self):
            return self.id

class CollegeStudent(Person, Student):
        def __init__(self, name, age, id):
            Person.__init__(self, name, age)
            Student.__init__(self, id)

obj = CollegeStudent('Kim', 22, '100036')
obj.show_info()
print(obj.get_id())
```

# 메소드 오버라이딩
⚫
“자식 클래스의 메소드가 부모 클래스의 메소드를 오버라이드(재정의)한다”고 말한다.
⚫ 다형성 : 함수 이름은 같으나 함수가 소속된 클래스에 따라 다른 코드가 수행되는 것 (C는 불가능, 객체
지향은 가능)
⚫ 메소드 오버라이드도 다형성의 일종임의 일종 : 수퍼클래스에서 정의된 함수를 서브클래스에서 재정의

```python
import math
class Shape:
    def __init__(self):
        pass
    def draw(self):
        print("draw()가 호출됨")
    def get_area(self):
        print("get_area()가 호출됨")
class Circle(Shape):
    def __init__(self, radius=0):
        super().__init__()
        self.radius = radius

    def draw(self):
        print("원을 그립니다.")

    def get_area(self):
        return math.pi * self.radius ** 2
c = Circle(10)
c.draw()
print("원의 면적:", c.get_area())
```
