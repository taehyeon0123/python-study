# 12장 상속

상속에 대한 이론과 학습

1. 캡슐화 : 데이터와 연산을 하나의 덩어리로 묶어서 다룸
➢ 정보은닉을 위한 가시성 : public (공개), protected (보호), private(숨김)
◼ 다른 클래스에서 이 클래스의 데이터와 연산을 접근하거나 접근불가로 만듬
◼ Public: 모든 클래스에서 접근가능, protected: 자식만 접근가능, private : 접근 불가
2. 상속 : 부모클래스를 자식클래스가 그대로 물려받아 추가적인 코드를 작성할 수 있음
➢ 다만, 부모의 private 한 것은 상속은 받으나 접근 불가능
◼ Protected는 외부 접근 불가능하나 자식이 물려받아 사용가능
3. 다형성 : 함수 이름이 같더라도 문맥(혹은 클래스)에 따라 다른 함수가 수행됨
➢ A.do(), B.do() 처럼 do()함수 이름은 같으나 클래스가 다르면 다른 함수
➢ B가 A를 상속하고 재정의 할 경우 동적 바인딩이 가능
