# map() 함수는 반복가능한 객체(리스트, 튜플 등)의 각 항목에 주어진 함수를 적용한 후, 결과를 반환한다

```python
def square(n):
    return n*n
mylist = [1, 2, 3, 4, 5]
result = list(map(square, mylist))
print(result)
```

# filter() 함수

filter() 함수는 특정 조건을 만족하는 요소만을 뽑는다.

```python
def myfilter(x):
    return x > 3
result = filter(myfilter, (1, 2, 3, 4, 5, 6))
print(list(result))
t1 = tuple(range(5, 10))
result2 = filter(lambda x: x > 3, t1)
print(list(result2))
```

# zip() 함수
⚫ zip() 함수는 2개의 시퀀스 자료형을 하나로 묶어주는 함수이다.

```python
numbers = [1, 2, 3, 4]
slist = ['one', 'two', 'three', 'four'
, 'five']
tuple1 = ('one', 'two', 'three', 'four', 'five')
set1 = {'one', 'two', 'three', 'four', 'five'}
dict1 = {0: "영", 1: "일"}
list(zip(numbers, slist))
[(1, 'one'), (2, 'two'), (3, 'three'), (4, 'four')]
```

# key 매개변수
정렬을 하다 보면 정렬에 사용되는 키를 개발자가 변경해주어야 하는 경우가 종종 있다.

```python
sorted("The health know not of their health, but only the sick".split(), key=str.lower)
['but', 'health', 'health,', 'know', 'not', 'of', 'only', 'sick', 'The', 'the', 'their']
```

```python
class Student:
    def __init__(self, name, grade, number):
        self.name = name
        self.grade = grade
        self.number = number

    def __repr__(self):
        return repr((self.name, self.grade, self.number))

students = [
Student('홍길동'
, 3.9, 20160303),
Student('김철수'
, 3.0, 20160302),
Student('최자영'
, 4.3, 20160301),
]
print(sorted(students, key=lambda student: student.number) )
```

# 오름차순 정렬과 내림차순 정렬

```python
sorted(students, key=lambda student: student.number, reverse=True)
[('홍길동'
, 3.9, 20160303), ('김철수'
, 3.0, 20160302), ('최자영'
, 4.3, 20160301)]
```
