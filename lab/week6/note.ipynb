# private 멤버 부모 클래스

```python
class Parent(object):
    def __init__(self):
        self.__money = 100

class Child(Parent):
        def __init__(self):
            super().__init__()
            self.__money = 200
        def get_money(self):
            return self.__money

obj = Child()
print(obj.get_money())  # This will raise an AttributeError since __money is private in both classes
```

# 다중 상속

```python
class Person:
    def __init__(self, name, age):
            self.name = name
            self.age = age

    def show_info(self):
        print(self.name, self.age)

class Student:
        def __init__(self,id):
            self.id = id
        def get_id(self):
            return self.id

class CollegeStudent(Person, Student):
        def __init__(self, name, age, id):
            Person.__init__(self, name, age)
            Student.__init__(self, id)

obj = CollegeStudent('Kim', 22, '100036')
obj.show_info()
print(obj.get_id())
```

# 메소드 오버라이딩
⚫
“자식 클래스의 메소드가 부모 클래스의 메소드를 오버라이드(재정의)한다”고 말한다.
⚫ 다형성 : 함수 이름은 같으나 함수가 소속된 클래스에 따라 다른 코드가 수행되는 것 (C는 불가능, 객체
지향은 가능)
⚫ 메소드 오버라이드도 다형성의 일종임의 일종 : 수퍼클래스에서 정의된 함수를 서브클래스에서 재정의

```python
import math
class Shape:
    def __init__(self):
        pass
    def draw(self):
        print("draw()가 호출됨")
    def get_area(self):
        print("get_area()가 호출됨")
class Circle(Shape):
    def __init__(self, radius=0):
        super().__init__()
        self.radius = radius

    def draw(self):
        print("원을 그립니다.")

    def get_area(self):
        return math.pi * self.radius ** 2
c = Circle(10)
c.draw()
print("원의 면적:", c.get_area())
```
